## Parte 1: Introducción a NumPy 1
### Temas de la clase
- Conversión de objetos enlistados a arreglos

### Tipos de objetos en Python

```python
print("1 + 1 =", 1 + 1)
print("Tipo de 1 + 1:", type(1 + 1))
print("Tipo de 1 + 1 + 0.5:", type(1 + 1 + 0.5))

x = "Hola a todos"
print(x)
print("Tipo de x:", type(x))
```

### Listas de elementos
Una lista es una enumeración de elementos. Para crear una lista se utilizan corchetes.

```python
l1 = []
print("Lista vacía:", l1)
print("Tipo de l1:", type(l1))

l1 = [1, 2, 3, 4, 5]
l2 = [6, 7, 8, 9, 10]

print("Lista 1:", l1)
print("Lista 2:", l2)
print("Longitud de l1:", len(l1))
print("Longitud de l2:", len(l2))
print("¿Tienen la misma longitud?", len(l1) == len(l2))
```

### Concatenación de listas
Al utilizar el signo `+` con listas, Python las concatena. Todavía no realiza una suma elemento por elemento.

```python
print("Listas concatenadas:", l1 + l2)

l3 = [1]
l4 = [100, 101]

print("l3 + l4:", l3 + l4)
```

### Acceso a elementos de una lista
Python comienza a contar los elementos desde el índice 0.

```python
print("Elemento con índice 0:", l1[0])
print("Elemento con índice 1:", l1[1])
print("Elemento con índice 2:", l1[2])
print("Elemento con índice 3:", l1[3])
print("Elemento con índice 4:", l1[4])
```

### Suma manual de listas elemento por elemento

```python
l = [
    l1[0] + l2[0],
    l1[1] + l2[1],
    l1[2] + l2[2],
    l1[3] + l2[3],
    l1[4] + l2[4]
]

print("Suma manual:", l)
```

### Rangos y ciclos

```python
print("Tipo de range:", type(range(0, 5)))

print("Números del rango:")
for i in range(0, 5):
    print(i)
```

### Suma de listas utilizando un ciclo

```python
l = [l1[i] + l2[i] for i in range(0, 5)]
print("Suma utilizando un ciclo:", l)
```

### Multiplicación de una lista
Al multiplicar una lista por 2, Python repite la lista (no la multiplica elemento por elemento).

```python
print("Lista repetida dos veces:", 2 * l2)
```

### Multiplicación elemento por elemento

```python
m = [0.5 * l2[i] for i in range(0, 5)]
print("Lista multiplicada por 0.5:", m)

m = [2 * l2[i] for i in range(0, 5)]
print("Lista multiplicada por 2:", m)
```

### Arreglos
NumPy permite convertir listas en arreglos. Los arreglos permiten realizar operaciones vectoriales.

```python
import numpy as np

l1_arreglo = np.array(l1)
l2_arreglo = np.array(l2)
l3_arreglo = np.array(l3)
l4_arreglo = np.array(l4)

print("Arreglo 1:", l1_arreglo)
print("Arreglo 2:", l2_arreglo)
print("Tipo de l1_arreglo:", type(l1_arreglo))
```

### Operaciones con arreglos
A diferencia de las listas, con arreglos sí se realizan operaciones vectoriales directas.

```python
print("Suma de arreglos:", l1_arreglo + l2_arreglo)
print("Arreglo 1 multiplicado por 2:", 2 * l1_arreglo)
print("Arreglo 2 multiplicado por 0.5:", 0.5 * l2_arreglo)
print("Suma de l3_arreglo y l4_arreglo:", l3_arreglo + l4_arreglo)
```

### Matrices
Una matriz tiene renglones y columnas. `reshape` permite cambiar la forma de un arreglo.

```python
lista = [i for i in range(10, 30)]

print("Lista del 10 al 29:", lista)
print("Cantidad de elementos:", len(lista))

array = np.array(lista)

print("Arreglo:", array)
```

```python
# Matriz de 10 renglones y 2 columnas
matriz1 = array.reshape(10, 2)
print("Matriz de 10 x 2:")
print(matriz1)
```

```python
# Matriz de 2 renglones y 10 columnas
matriz2 = array.reshape(2, 10)
print("Matriz de 2 x 10:")
print(matriz2)
```

```python
# Matriz de 5 renglones y 4 columnas
matriz3 = array.reshape(5, 4)
print("Matriz de 5 x 4:")
print(matriz3)
```

## Parte 2: Distribución muestral y Teorema del Límite Central

Aquí se simula una población, se toman muestras de distintos tamaños y se observa cómo se distribuyen las medias muestrales — la base de la estadística inferencial: aunque la población no sea normal, la distribución de las medias muestrales tiende a serlo conforme crece el tamaño de muestra.

### 1. Crear una población

```python
import numpy as np
import matplotlib.pyplot as plt

poblacion = np.random.uniform(0, 100, 10000)

plt.hist(poblacion, bins=30, edgecolor='black')
plt.title("Distribución de la población")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.show()
```

### 2. Tomar 1000 muestras de tamaño 30 y calcular la media de cada muestra

```python
medias = []

for i in range(1000):
    muestra = np.random.choice(poblacion, size=30)
    medias.append(np.mean(muestra))

plt.hist(medias, bins=30, edgecolor='black')
plt.title("Distribución de las medias muestrales")
plt.xlabel("Media")
plt.ylabel("Frecuencia")
plt.show()
```

### 3. Comparar diferentes tamaños de muestra
Entre más grande el tamaño de muestra (n), más se "achina" y se parece a una normal la distribución de las medias.

```python
tamanos = [5, 30, 100]

for n in tamanos:
    medias = []

    for i in range(1000):
        muestra = np.random.choice(poblacion, size=n)
        medias.append(np.mean(muestra))

    plt.hist(medias, bins=30, edgecolor='black')
    plt.title(f"Medias muestrales con n = {n}")
    plt.xlabel("Media")
    plt.ylabel("Frecuencia")
    plt.show()
```
